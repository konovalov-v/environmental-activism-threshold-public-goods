# Theoretical figures

This notebook provides a guided route through the four theoretical figures in the paper. It explains the quantities being compared, runs the figure script, and displays the resulting images.

No survey data are needed. The calculations are deterministic and use a fixed grid of contribution probabilities.

## 1. Set up paths

The next cell finds the repository root and defines the location of the generated figures.

```python
from pathlib import Path
import subprocess
import sys

from IPython.display import Image, display


def find_repository_root(start=Path.cwd()):
    for candidate in (start, *start.parents):
        if (candidate / "run_all.py").exists() and (candidate / "code").is_dir():
            return candidate
    raise FileNotFoundError("Open this notebook from inside the replication repository.")


ROOT = find_repository_root()
FIGURES = ROOT / "output" / "figures"
print(f"Repository: {ROOT}")
```

## 2. Quantities shown in the figures

Let $p$ be the probability that one person contributes to the public good.

When the participation threshold $k$ is known, the pivotal probability is

$$q(p)=\binom{n-1}{k-1}p^{k-1}(1-p)^{n-k}.$$

It is the probability that exactly $k-1$ of the other $n-1$ people contribute, so one person's contribution is decisive.

When the threshold is uncertain, the corresponding quantity is

$$\omega(p)=F_\theta\bigl(p(n-1)+1\bigr)-F_\theta\bigl(p(n-1)\bigr),$$

where $F_\theta$ describes uncertainty about the threshold.

The figures compare these benefits with $G^{-1}(p)$, the cost level below which a share $p$ of people would be willing to contribute. A crossing indicates an internal equilibrium contribution rate.

## 3. Generate the figures

The repository script evaluates the formulas on 100 evenly spaced values from 0 to 1 and saves all four figures. It does not draw random numbers.

```python
subprocess.run([sys.executable, str(ROOT / "code" / "05_figures_theory.py")], check=True)
```

## 5. Uncertain participation threshold

Figure 3 changes the expected threshold while holding the amount of uncertainty fixed. This shows how the expected coordination requirement shifts the possible equilibrium contribution rates.

Figure 4 holds the expected threshold fixed and changes the amount of uncertainty. This isolates how less or more precise beliefs about the threshold affect the possible equilibrium contribution rates.

```python
display(Image(filename=str(FIGURES / "fig4_threshold_uncertain_varying_alpha.png"), width=1100))
```
